```python
import pandas as pd
from pathlib import Path
```

```python
carpetas = {
    Path('..') / 'data' / 'closed_eye': 0,
    Path('..') / 'data' / 'open_eye': 1
}
dataset = []
```

```python
for carpeta, etiqueta in carpetas.items():
    if carpeta.exists():
        for archivo in carpeta.iterdir():
            if archivo.suffix.lower() in ('.png', '.jpg', '.jpeg'):
                dataset.append({
                    'filepath': str(archivo),
                    'label': etiqueta
                })
```

```python
df = pd.DataFrame(dataset)
```

```python
df.head(5)
```

```python
df = df.sample(frac=1, random_state=42)
```

```python
df.to_csv('dataset.csv', index=False)
```
